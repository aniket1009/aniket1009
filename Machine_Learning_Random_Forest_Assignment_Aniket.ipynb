# Task 6 Questions:

1. What is accuracy?
* Answer: Accuracy is the percentage of correctly predicted samples out of the total number of samples.
   
  Formula:

           Accuracy = (Correct Predictions) / (Total Predictions)

2. What is Precision?
* Answer:
    Precision measures how many of the predicted positive instances are actually correct.

   Formula:
           
           Precision = TP / (TP + FP)

3. What is Recall?
* Answer:
Recall measures how many actual positive instances are correctly identified by the model.

  Formula:

          Recall = TP / (TP + FN)

  where:

          FN = False Negatives

4. What is F1 Score?
* Answer:
F1 Score is the harmonic mean of Precision and Recall. It provides a balance between Precision and Recall.

  Formula:

          F1 Score = 2 × (Precision × Recall) / (Precision + Recall)

5. Why is Accuracy not always enough?
* Answer: Accuracy alone can be misleading, especially when the dataset is imbalanced. A model may achieve high accuracy but still perform poorly on certain classes. Therefore, Precision, Recall, and F1 Score should also be considered.

```python
tree_values = [10, 50 ,100, 200, 500]
accuracies = []

for n in tree_values:
  model = RandomForestClassifier(
      n_estimators= n,
      random_state = 42
  )
  model.fit(x_train, y_train)
  preds = model.predict(x_test)
  acc = accuracy_score(y_test, preds)
  accuracies.append(acc)

  print(f"Trees: {n} ---> Accuracy: {acc:.4f}  ")
```

# Task 13: Out-of-Bag (OOB) Score

1. What is an OOB sample?
* Answer:
An Out-of-Bag (OOB) sample is a data sample that is not selected during bootstrap sampling for training a particular decision tree.

2. Why can OOB score act as an internal validation score?
* Answer:
Since OOB samples are not used to train a tree, they can be used to test that tree. Therefore, the OOB score provides an internal estimate of model performance without requiring a separate validation dataset.

3. Approximately what percentage of samples become OOB?
* Answer:
Approximately 36.8% of the samples become Out-of-Bag samples during bootstrap sampling.

4. What are the advantages of OOB validation?
* Answer:
  1. It eliminates the need for a separate validation dataset.
  2. It makes efficient use of all available data.
  3. It provides an unbiased estimate of model performance.
  4. It reduces the need for cross-validation.

#Observations:

    OOB Score = 0.9699

# Interpretation:

* The Random Forest model achieved an OOB score of 96.99%, indicating excellent predictive performance. The high OOB score suggests that the model generalizes well to unseen data.


```python
from sklearn.preprocessing import StandardScaler

# Without Scaling
model_without = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model_without.fit(x_train, y_train)

pred_without = model_without.predict(x_test)

acc_without = accuracy_score(y_test, pred_without)

# With Scaling
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

model_with = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model_with.fit(x_train_scaled, y_train)

pred_with = model_with.predict(x_test_scaled)

acc_with = accuracy_score(y_test, pred_with)

print("Without Scaling Accuracy:", acc_without)
print("With Scaling Accuracy:", acc_with)
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

housing = fetch_california_housing()

x_reg = housing.data
y_reg = housing.target

print("Dataset Shape:", x_reg.shape)
```

```python
x_train_reg, x_test_reg, y_train_reg, y_test_reg = train_test_split(
    x_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("X_train Shape:", x_train_reg.shape)
print("X_test Shape:", x_test_reg.shape)
```

```python
rf_reg = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_reg.fit(x_train_reg, y_train_reg)

print("Regression Model Trained Successfully!")
```

```python
pred_reg = rf_reg.predict(x_test_reg)

mae = mean_absolute_error(y_test_reg, pred_reg)

rmse = mean_squared_error(
    y_test_reg,
    pred_reg
) ** 0.5

r2 = r2_score(y_test_reg, pred_reg)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

# Final Theory Questions

Q1. What is Random Forest?

Random Forest is an ensemble machine learning algorithm that consists of multiple decision trees. It combines the predictions of these trees to produce a final prediction. For classification, it uses majority voting, while for regression, it uses averaging.

Q2. Why is Random Forest considered an Ensemble Learning algorithm?

Random Forest is considered an Ensemble Learning algorithm because it combines the predictions of multiple decision trees to improve accuracy, stability, and generalization performance.

Q3. Explain Bagging in detail.

Bagging (Bootstrap Aggregating) is an ensemble learning technique in which multiple training datasets are created using bootstrap sampling. A separate decision tree is trained on each dataset. The predictions from all trees are then combined using majority voting (for classification) or averaging (for regression). Bagging reduces variance and helps prevent overfitting.

Q4. What is Bootstrap Sampling?

Bootstrap Sampling is a sampling technique in which random samples are selected from the original dataset with replacement. Since sampling is performed with replacement, the same sample may appear multiple times in a bootstrap sample.

Q5. What is Feature Randomness?

Feature Randomness refers to the process of selecting a random subset of features at each split while building decision trees in a Random Forest. This increases diversity among trees and improves model performance.

Q6. Explain the role of n_estimators.

The parameter n_estimators specifies the number of decision trees in the Random Forest. Increasing the number of trees generally improves model performance, but it also increases computational cost and training time.

Q7. What is the purpose of max_depth?

The max_depth parameter controls the maximum depth of each decision tree. It helps prevent overfitting by limiting how deep the trees can grow.

Q8. What is OOB Score?

Out-of-Bag (OOB) Score is an internal validation score computed using samples that were not selected during bootstrap sampling. It provides an estimate of model performance without requiring a separate validation dataset.

Q9. What is Feature Importance?

Feature Importance is a measure that indicates how much each feature contributes to the prediction made by the model. Features with higher importance scores have a greater influence on the model's decisions.

Q10. Why does Random Forest generally perform better than a single Decision Tree?

Random Forest generally performs better because it combines the predictions of multiple decision trees, reducing variance and overfitting. This leads to more accurate and stable predictions.

Q11. What are the advantages of Random Forest?

1. High predictive accuracy.
2. Reduces overfitting compared to a single decision tree.
3. Handles both classification and regression problems.
4. Works well with large datasets.
5. Provides feature importance scores.
6. Is robust to noise and outliers.

Q12. What are the disadvantages of Random Forest?

1. Requires more computational resources.
2. Training can be slow for very large datasets.
3. Consumes more memory due to multiple trees.
4. Models are less interpretable compared to a single decision tree.
5. Can be computationally expensive for real-time applications.
